### Datos etiquetados.

- Tipo de algoritmo de Machine Learning que trabajan con datos "etiquetados". ¿Esto qué significa? El algoritmo tiene un histórico de datos (features), cuyo target es conocido.

- ¿El algoritmo sabe lo que es un perro y lo que es un gato? NO. Detecta patrones y clasifica los datos en dos grupos. Después somos nosotros los encargados de interpretar esos grupos.

### ¿Que es un modelo?

Un modelo es un conjunto de fórmulas matemáticas que expresan relaciones y patrones entre variables, utilizados para estudiar comportamientos de sistemas complejos ante situaciones difíciles de observar en la realidad.

Es una manera de traducir a la matemática relaciones entre datos de la vida real.

Con un modelo puedo predecir cierto output, como por ejemplo si el paciente contrae o no cáncer de pulmón, a partir de un conjunto de inputs, que podrían ser edad, fumador, enfermedades previas…

### ¿Como se crea un modelo?

Un modelo necesita un histórico de datos que le sirvan de muestra, al igual que el aprendizaje de un ser humano. Con esos datos aprende patrones para predecir el output (tiene o no cáncer), de manera que cuando vengan inputs nuevos, sabrá dar un output.

### Target y Features.

Esto es pura terminología de Machine Learning.

- Target: Se trata de la variable objetivo, la variable a predecir, el output del modelo. Por norma general es una única variable, ya sea numérica o categórica.

- Features: Todo el resto de variables utilizadas como input del modelo.

### Objetivo de un modelo. (Generalizar ante nuevos datos)

- Posible problema: El modelo no generaliza bien

- Solución: Dividir en train/test. Guardo unos datos de test, ajenos completamente al entrenamiento. Y pruebo el modelo entrenado, a ver qué tal generaliza con datos nuevos.

### División Train y Test.

- ¿Cuántos datos guardo para test? Entre un 15-30 %

- Entrenamos varios modelos (CON EL ...)

- ¿Objetivo?

1. Elegir el modelo con menor cantidad de errores.

2. Que funcione bien en test, que generalice.



### Posibles problemas:

- Overfitting o sobreajuste: Sobreajuste de un modelo a los datos de entrenamiento. Como consecuencia el modelo no va a generalizar bien. No se va a comportar como debería cuando le entren nuevos datos.

    - Estoy ajustándome demasiado a mis problemas de entrenamiento y no voy a poder generalizar/resolver nuevos problemas, desconocidos para mí.

    - Tengo que aprender los patrones de los problemas, y la manera de resolverlos, no aprendérmelos de memoria. (ejemplo estudiar examen de mates y comparación con modelo)

    - ¿Solución?
        - Utilizar más datos en training
        - Reducir la complejidad de los modelos
        - Cross Validation!!

- Underfitting: Se produce cuando entrenamos con pocos datos un modelo, y este no es capaz de identificar los patrones que hay en los datos para sus predicciones. Por tanto este modelo NO puede generalizar bien, para cuando le lleguen datos nuevos.

### Otros factores a tener en cuenta:

- Bias: Capacidad de un modelo de ajustarse a los datos de entrenamiento. Si se ajusta de más, tendremos Overfitting, y por tanto un Variance alto.

- Variance: Capacidad de un modelo para mantener sus regiones de decisión ante pequeñas variaciones de los datos.



        - Objetivo: Reducir el bias y el variance al maximo. Con esto conseguimos tener un modelo que se ajuste al patrón de los datos y al mismo tiempo generalizar con datos futuros

### Técnica para solventar esto.

- Regularization (simplifica el modelo para que generalice bien): 
    - ¿Qué conseguimos con esto?: Aumentamos los errores en train a cambio de disminuirlos en test. Reducimos ese gap entre errores

- Cross Validation (utilizamos los datos de train y test y viceverssa):

    - Ejemplo para K=5
        - Mezclar datos
        - Divide tus datos en 5 conjuntos (A, B, C, D, E)
        - Guarda 1/5  de los datos para test. Entrena con B, C, D, E y utiliza A para test
        - Te da un error
        - Ahora guárdate B para test y utiliza A, C, D, E de entrenamiento.
        - Da otro error
        - Cuando acabes de hacer eso con todos los folds, calcula la media de todos los erros obtenidos y será el error cometido por el modelo.

    - ¿Qué conseguimos con esto?: Garantizar que la evaluación del modelo sea independiente de la partición. Al utilizar todos los datos de train, en test, y los datos de test a su vez en train también, conseguiré que mi modelo generalice bien, y por tanto no me produzca overfitting.
    
    - ¿En cuántos k-folds hay que dividirlo?: Se suelen usar 10, aunque depende del volumen del dataset. Si es muy grande, dividir en 5 k-folds.

### Consideraciones:

- Cuidado al aplicar modificaciones en los conjuntos de train/test. No podemos contaminar ambos conjuntos con datos del otro. El conjunto de test se separa de los datos y no se toca hasta el final. Todas las transformaciones que apliquemos en train, tiene que salir únicamente del conjunto de train.

- Ejemplo: si vamos a imputar missings en una columna por su media (imaginemos 31.24 con todos los datos), y sustituimos en train, estaríamos utilizando los datos de test para sustituir ese missing en train, cuando en realidad para los datos de train, los de test son completamente desconocidos. La media del conjunto de train podría ser de 22.67, muy lejos de la media de train + test.

- ¿Cómo aplicar las transformaciones/limpieza obtenidas hasta ahora?

    - Scalers: por ejemplo, si tenemos un StandardScaler en train, usar ese mismo StandardScaler en test.
    - Missings: si aplicábamos la media/mediana/moda en train, aplicar esa misma métrica en test.
    - Feature reduction: eliminar las features que quitábamos en train (y añadir los dummies vacíos en test)
    - Feature engineering: mismos cálculos que en train.

## 3. Coeficiente de Determinación (R^2)

El Coeficiente de Determinación (R^2) es una métrica que indica la proporción de la varianza en la variable dependiente que es explicada por el modelo. Se define como:



El valor de R^2 varía entre 0 y 1, donde 1 indica que el modelo explica completamente la variabilidad de los datos. Un R^2 negativo puede aparecer si el modelo es peor que simplemente predecir el valor promedio de los datos.


```python
# Ejemplo métricas.

from sklearn . metrics import mean_absolute_error , mean_squared_error , r2_score
import numpy as np

# Simulacion de etiquetas verdaderas y predicciones del modelo
y_true = np. array ([3.5 , -0.5, 2.0 , 7.0])
y_pred = np. array ([2.5 , 0.0 , 2.1 , 7.8])

# Calculo de metricas
mae = mean_absolute_error (y_true , y_pred )
mse = mean_squared_error (y_true , y_pred )
rmse = np. sqrt (mse )
r2 = r2_score ( y_true , y_pred )

print (f'MAE: {mae :.2f}')
print (f'MSE: {mse :.2f}')
print (f'RMSE : { rmse :.2f}')
print (f'R^2: {r2 :.2f}')
```

- MAE: Mide el error absoluto promedio, que es 0.48.

- MSE: El error cuadrático medio, que es 0.57, penaliza los errores más grandes.

- RMSE: La raíz del MSE es 0.75, lo que permite interpretar el error en las mismas unidades que las predicciones.

- R^2: El coeficiente de determinaci´on es 0.99, lo que indica que el modelo explica casi toda la varianza de los datos.

## 2. K vecinos más cercanos (KNN)
El algoritmo de K Vecinos Más Cercanos (KNN) es uno de los algoritmos de clasificación más simples pero efectivos en el aprendizaje automático supervisado. Se clasifica dentro de los algoritmos de Lazy Learning porque no entrena un modelo explícito durante la fase de entrenamiento, sino que realiza predicciones utilizando directamente el conjunto de datos de entrenamiento en cada consulta nueva.

### 2.1. Fundamentos del Algoritmo
El algoritmo KNN se basa en la idea de que instancias similares (cercanas en el espacio de características) deben tener la misma etiqueta de clase. Para clasificar una nueva instancia (un nuevo dato), el algoritmo busca los k puntos más cercanos (vecinos) en el conjunto de entrenamiento y asigna la clase más frecuente entre ellos.

### 2.2. Elección del Valor de k
Uno de los parámetros que tenemos que asignar al modelo es el valor de k. El valor de k define cuántos vecinos se consideran para tomar la decisión de si un punto pertenece a una clase o a otra. Si k es muy pequeño, el algoritmo puede ser sensible al ruido; si es muy grande, puede perder precisión al promediar puntos que no son muy relevantes.
Imaginemos que tenemos un conjunto de puntos y queremos predecir la clase
de un nuevo punto:

- Si k = 1, el nuevo punto se clasificará de acuerdo con el vecino más cercano (será de la clase del punto que esté a menor distancia de él).

- Si k = 5, se contarán las clases de los 5 vecinos más cercanos y se seleccionará la clase mayoritaria.

En este caso es importante que escojamos un valor impar del parámetro k, puesto que de lo contrario podría darse un empate. Por tanto hemos visto que la elección del valor de k es crucial para el rendimiento del algoritmo KNN.

### 2.3. Preprocesamiento de los Datos
Para que el KNN funcione de manera óptima, es esencial realizar un preprocesamiento adecuado:

- Normalización de los datos: Dado que KNN depende de la distancia, es fundamental que los atributos estén en la misma escala. Esto se puede lograr utilizando técnicas como la normalización o estandarización.

- Eliminación de características irrelevantes: Dado que cada atributo contribuye al cálculo de la distancia, los atributos irrelevantes pueden introducir ruido y afectar la precisión del modelo.

### 2.4. Algoritmo de KNN Paso a Paso
El algoritmo KNN sigue estos pasos:

1. Elegir número de vecinos: Selecciona el número de vecinos cercanos, k.

2. Calcular Distancias: Calcula la distancia entre el nuevo punto y todos los puntos del conjunto de entrenamiento. Tener en cuenta que nosotros previamente tendremos que haber definido al modelo qué tipo de distancia tiene que utilizar.

3. Seleccionar Vecinos: Identifica los k puntos más cercanos.

4. Asignar Clase: La clase más frecuente entre los k vecinos se asigna al nuevo punto.

#### 2.8.2. La Maldición de la Dimensionalidad
Uno de los principales problemas del algoritmo KNN es lo que se conoce como la maldición de la dimensionalidad. A medida que aumenta el número de características o dimensiones en los datos, la distancia entre los puntos de datos tiende a aumentar, lo que hace que sea más difícil identificar correctamente a los vecinos más cercanos.

#### 2.8.3. Impacto en el KNN
En un espacio de alta dimensionalidad, todos los puntos de datos parecen estar igualmente distantes entre sí. Esto reduce la capacidad del algoritmo KNN para hacer distinciones útiles entre los datos, lo que degrada su rendimiento.
Para mitigar este problema, es recomendable usar técnicas de reducción de dimensionalidad como el Análisis de Componentes Principales (PCA) o seleccionar únicamente las características más relevantes.

### 2.9. A tener en cuenta
El algoritmo KNN es un algoritmo simple pero efectivo para tareas de clasificación y regresión. Sin embargo, debido a su dependencia de la distancia, es crucial elegir adecuadamente el valor de k, preprocesar los datos y seleccionar las métricas adecuadas para asegurar un rendimiento óptimo. Su simplicidad lo hace muy popular en aplicaciones prácticas, aunque para grandes conjuntos de datos o en espacios de alta dimensionalidad, puede que no sea la mejor opción.
